 # use a display method to print the attributes of the objects data using self keyword

```python
class Student:
    name="Abhijeet"
    age=23
    
    def display(self):
        print(self.name)
        print(self.age)
        
s1=Student()
s1.display()  

s2=Student()
s2.name="ANK"
s2.age=24 

s2.display()    
```

# used another method to get info of attributes from the user 

```python
class Student:
    name=""
    age=""
    
    def get_info(self):
        self.name=input("Enter the Name")
        self.age=int(input("Enter the Age"))
    
    def display(self):
        print(f"my name is {self.name} and my age is {self.age}")
        
        
s1=Student()
s1.get_info()
s1.display()  

s2=Student()
s2.get_info()
s2.display()    
```

# We used a parameter inside a method 

```python
class Student:
    name=""
    age=""
    
    def get_info(self,name,age):
        self.name=name
        self.age=age
    
    def display(self):
        print(f"my name is {self.name} and my age is {self.age}")
        
        
s1=Student()
s1.get_info("abhi",22)
s1.display()  

s2=Student()
s2.get_info("abhi",24)
s2.display()    
```

# A constructor is a special method that automatically runs when you create an object.
# We use a constructor to automatically set the required information when an object is created, so the object is ready to use immediately.

```python
class Student:
    
    def __init__(self, name , age):
        print("this is a constructor")
        self.name=name
        self.age=age
    
    def display(self):
            print(f"my name is {self.name} , my age is {self.age}")
        
s1=Student("abhijeet",21)
s1.display()

    
```

```python
class Animal():
    
    def __init__(self,name,age):
        self.name=name
        self.age=age
        
    def Bark(self):
        print("Barking")
    def eat(self):
        print("Eating")
        
class Dog(Animal):
    def __init__(self, name, age, breed):  
        super().__init__(name, age)  #is used to call the constructor of the parent class.
        self.breed=breed
        
    def display(self):
        print(f"name is {self.name}, age is {self.age} and the breed is {self.breed}")
        

dog=Dog("Luffy","7","pirate")

dog.Bark() 
dog.eat()  
dog.display()     
```

# Encapsulation

```python
class Bank :
    
    def __init__(self,name,__balance):
        self.name= name
        self.__balance=__balance
        
    def getbal(self):
        return self.__balance
    
    def setbal(self, new_balance):
        self.__balance+= new_balance
        return self.__balance
    
    
acc=Bank("Abhijeet",1000)
acc.getbal()
print(acc.setbal(200))
    
    
```
